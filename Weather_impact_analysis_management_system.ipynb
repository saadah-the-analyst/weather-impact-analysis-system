```python
"""
WEATHER IMPACT ANALYSIS SYSTEM
ClimateSmart Analytics Africa - Junior Data Analyst Project

This program analyses simulated weather observations from Lagos, Kano, and
Ibadan to explore possible links between weather conditions, agricultural
outcomes, and operational disruption.

IMPORTANT DATA DISCLAIMER: This is a simulated educational dataset. It is
NOT official Nigerian meteorological data and should not be presented as such.

Built using only Python fundamentals (no Pandas, NumPy, Matplotlib, APIs,
CSV libraries, or databases), as required for this stage of the project.
"""

# ================================================================
# SECTION 1: THE DATASET
# ================================================================

weather_data = [
    {"date": "2025-04-01", "location": "Lagos", "temperature": 31, "rainfall": 12,
     "humidity": 78, "wind_speed": 14, "crop_yield": 4.2, "disruption": "No"},
    {"date": "2025-04-05", "location": "Lagos", "temperature": 32, "rainfall": 18,
     "humidity": 81, "wind_speed": 16, "crop_yield": 4.0, "disruption": "No"},
    {"date": "2025-04-10", "location": "Lagos", "temperature": 33, "rainfall": 45,
     "humidity": 85, "wind_speed": 19, "crop_yield": 3.9, "disruption": "Yes"},
    {"date": "2025-04-15", "location": "Lagos", "temperature": 30, "rainfall": 67,
     "humidity": 88, "wind_speed": 21, "crop_yield": 3.7, "disruption": "Yes"},
    {"date": "2025-04-20", "location": "Lagos", "temperature": 29, "rainfall": 92,
     "humidity": 90, "wind_speed": 23, "crop_yield": 3.5, "disruption": "Yes"},
    {"date": "2025-05-01", "location": "Lagos", "temperature": 31, "rainfall": 35,
     "humidity": 82, "wind_speed": 18, "crop_yield": 3.9, "disruption": "No"},
    {"date": "2025-05-05", "location": "Lagos", "temperature": 34, "rainfall": 8,
     "humidity": 74, "wind_speed": 15, "crop_yield": 3.6, "disruption": "No"},
    {"date": "2025-05-10", "location": "Lagos", "temperature": 35, "rainfall": 5,
     "humidity": 72, "wind_speed": 17, "crop_yield": 3.4, "disruption": "No"},
    {"date": "2025-05-15", "location": "Lagos", "temperature": 32, "rainfall": 58,
     "humidity": 86, "wind_speed": 20, "crop_yield": 3.8, "disruption": "Yes"},
    {"date": "2025-05-20", "location": "Lagos", "temperature": 30, "rainfall": 110,
     "humidity": 92, "wind_speed": 25, "crop_yield": 3.3, "disruption": "Yes"},

    {"date": "2025-04-01", "location": "Kano", "temperature": 36, "rainfall": 2,
     "humidity": 38, "wind_speed": 12, "crop_yield": 3.8, "disruption": "No"},
    {"date": "2025-04-05", "location": "Kano", "temperature": 37, "rainfall": 0,
     "humidity": 35, "wind_speed": 14, "crop_yield": 3.6, "disruption": "No"},
    {"date": "2025-04-10", "location": "Kano", "temperature": 38, "rainfall": 4,
     "humidity": 33, "wind_speed": 16, "crop_yield": 3.4, "disruption": "Yes"},
    {"date": "2025-04-15", "location": "Kano", "temperature": 39, "rainfall": 1,
     "humidity": 31, "wind_speed": 18, "crop_yield": 3.1, "disruption": "Yes"},
    {"date": "2025-04-20", "location": "Kano", "temperature": 40, "rainfall": 0,
     "humidity": 29, "wind_speed": 20, "crop_yield": 2.9, "disruption": "Yes"},
    {"date": "2025-05-01", "location": "Kano", "temperature": 37, "rainfall": 8,
     "humidity": 40, "wind_speed": 15, "crop_yield": 3.3, "disruption": "No"},
    {"date": "2025-05-05", "location": "Kano", "temperature": 39, "rainfall": 3,
     "humidity": 34, "wind_speed": 17, "crop_yield": 3.0, "disruption": "Yes"},
    {"date": "2025-05-10", "location": "Kano", "temperature": 41, "rainfall": 0,
     "humidity": 28, "wind_speed": 19, "crop_yield": 2.7, "disruption": "Yes"},
    {"date": "2025-05-15", "location": "Kano", "temperature": 38, "rainfall": 12,
     "humidity": 42, "wind_speed": 16, "crop_yield": 3.2, "disruption": "No"},
    {"date": "2025-05-20", "location": "Kano", "temperature": 40, "rainfall": 5,
     "humidity": 36, "wind_speed": 21, "crop_yield": 2.8, "disruption": "Yes"},

    {"date": "2025-04-01", "location": "Ibadan", "temperature": 30, "rainfall": 20,
     "humidity": 76, "wind_speed": 13, "crop_yield": 4.5, "disruption": "No"},
    {"date": "2025-04-05", "location": "Ibadan", "temperature": 31, "rainfall": 25,
     "humidity": 79, "wind_speed": 15, "crop_yield": 4.4, "disruption": "No"},
    {"date": "2025-04-10", "location": "Ibadan", "temperature": 30, "rainfall": 40,
     "humidity": 82, "wind_speed": 17, "crop_yield": 4.2, "disruption": "No"},
    {"date": "2025-04-15", "location": "Ibadan", "temperature": 29, "rainfall": 65,
     "humidity": 87, "wind_speed": 20, "crop_yield": 4.0, "disruption": "Yes"},
    {"date": "2025-04-20", "location": "Ibadan", "temperature": 28, "rainfall": 85,
     "humidity": 89, "wind_speed": 22, "crop_yield": 3.8, "disruption": "Yes"},
    {"date": "2025-05-01", "location": "Ibadan", "temperature": 30, "rainfall": 32,
     "humidity": 80, "wind_speed": 16, "crop_yield": 4.3, "disruption": "No"},
    {"date": "2025-05-05", "location": "Ibadan", "temperature": 32, "rainfall": 18,
     "humidity": 75, "wind_speed": 14, "crop_yield": 4.1, "disruption": "No"},
    {"date": "2025-05-10", "location": "Ibadan", "temperature": 33, "rainfall": 10,
     "humidity": 72, "wind_speed": 15, "crop_yield": 4.0, "disruption": "No"},
    {"date": "2025-05-15", "location": "Ibadan", "temperature": 31, "rainfall": 55,
     "humidity": 84, "wind_speed": 19, "crop_yield": 4.0, "disruption": "Yes"},
    {"date": "2025-05-20", "location": "Ibadan", "temperature": 29, "rainfall": 100,
     "humidity": 91, "wind_speed": 24, "crop_yield": 3.6, "disruption": "Yes"},
]


# ================================================================
# SECTION 2: SMALL HELPER FUNCTIONS
# (Used across several other functions, so they live up here.
#  Written as plain loops instead of comprehensions, to stay
#  within the allowed fundamentals for this stage of the project.)
# ================================================================

# Note: All custom functions in this module are defined and fully functional. 
# They have been designed and tested for the execution pipeline, though standalone function calls are omitted here.

def get_unique_locations(data):
    """Build a list of the distinct locations found in the dataset."""
    locations = []
    for record in data:
        if record["location"] not in locations:
            locations.append(record["location"])
    return locations


def is_extreme_event(record):
    """
    An observation counts as an extreme event if ANY of the following hold:
    rainfall > 100mm, temperature > 35°C, or wind speed > 25km/h.
    """
    return (record["rainfall"] > 100
            or record["temperature"] > 35
            or record["wind_speed"] > 25)


def get_month_name(date_string):
    """Convert a 'YYYY-MM-DD' date string into a month name, e.g. 'April'."""
    month_names = {
        "01": "January", "02": "February", "03": "March", "04": "April",
        "05": "May", "06": "June", "07": "July", "08": "August",
        "09": "September", "10": "October", "11": "November", "12": "December"
    }
    month_number = date_string[5:7]
    return month_names[month_number]


def get_unique_months(data):
    """Build a list of the distinct months represented in the dataset."""
    months = []
    for record in data:
        month_name = get_month_name(record["date"])
        if month_name not in months:
            months.append(month_name)
    return months


# ================================================================
# SECTION 3: DATASET SUMMARY (Task 5 / Basic Questions)
# ================================================================

def display_dataset_summary(data):
    """
    Print a high-level summary answering: how many records, how many
    locations, the earliest and latest dates, and which variables exist.
    """
    unique_locations = get_unique_locations(data)

    earliest_record = data[0]
    latest_record = data[0]
    for record in data:
        if record["date"] < earliest_record["date"]:
            earliest_record = record
        if record["date"] > latest_record["date"]:
            latest_record = record

    print("""
========================================
WEATHER DATASET SUMMARY
========================================""")
    print("Total Records:", len(data))
    print("Locations:", len(unique_locations), "-", unique_locations)
    print("Earliest Date:", earliest_record["date"])
    print("Latest Date:", latest_record["date"])
    print("Variables Available:", list(data[0].keys()))


# ================================================================
# SECTION 4: WHOLE-DATASET BASIC STATS (Tasks 6-8)
# ================================================================

def calculate_average_temperature(data):
    """Calculate the average temperature across the whole dataset."""
    total_temperature = 0
    count = 0
    for record in data:
        total_temperature += record["temperature"]
        count += 1
    return total_temperature / count


def calculate_total_rainfall(data):
    """Calculate the total rainfall across the whole dataset."""
    total_rainfall = 0
    for record in data:
        total_rainfall += record["rainfall"]
    return total_rainfall


def count_disruptions(data):
    """Count how many records had an operational disruption."""
    disruption_count = 0
    for record in data:
        if record["disruption"] == "Yes":
            disruption_count += 1
    return disruption_count


def find_highest_temperature(data):
    """Find the highest temperature recorded, with its date and location."""
    highest_temperature = 0
    highest_date = ""
    highest_location = ""
    for record in data:
        if record["temperature"] > highest_temperature:
            highest_temperature = record["temperature"]
            highest_date = record["date"]
            highest_location = record["location"]
    return highest_temperature, highest_date, highest_location


def find_highest_rainfall(data):
    """Find the highest rainfall recorded, with its date and location."""
    highest_rainfall = 0
    highest_date = ""
    highest_location = ""
    for record in data:
        if record["rainfall"] > highest_rainfall:
            highest_rainfall = record["rainfall"]
            highest_date = record["date"]
            highest_location = record["location"]
    return highest_rainfall, highest_date, highest_location


# ================================================================
# SECTION 5: WEATHER CLASSIFICATION (Tasks 9-10)
# ================================================================

def classify_rainfall(rainfall):
    """
    Classify a rainfall value (mm) using the project's thresholds:
    0 = No Rain | 1-10 = Light | 11-50 = Moderate | 51-100 = Heavy | >100 = Extreme
    """
    if rainfall == 0:
        return "No Rain"
    elif rainfall <= 10:
        return "Light"
    elif rainfall <= 50:
        return "Moderate"
    elif rainfall <= 100:
        return "Heavy"
    else:
        return "Extreme"


def classify_temperature(temperature):
    """
    Classify a temperature value (°C) using the project's thresholds:
    <25 = Cool | 25-30 = Moderate | 31-35 = Hot | >35 = Extreme Heat
    """
    if temperature < 25:
        return "Cool"
    elif temperature <= 30:
        return "Moderate"
    elif temperature <= 35:
        return "Hot"
    else:
        return "Extreme Heat"


# ================================================================
# SECTION 6: EXTREME EVENT DETECTION (Tasks 11-12)
# ================================================================

def display_extreme_events(data):
    """Print every record and flag whether it is an extreme event."""
    extreme_count = 0
    print("""
========================================
EXTREME WEATHER EVENTS
========================================""")
    for record in data:
        if is_extreme_event(record):
            extreme_count += 1
            print(f"{record['date']} | {record['location']} | "
                  f"{record['temperature']}°C | {record['rainfall']}mm | "
                  f"{record['wind_speed']}km/h")
    print(f"\nTotal Extreme Weather Events: {extreme_count}")
    return extreme_count


# ================================================================
# SECTION 7: LOCATION ANALYSIS (Tasks 13-15)
# ================================================================

def analyse_location(location, data):
    """
    Analyse all records for one location: record count, average and
    maximum temperature, total and maximum rainfall, disruption count,
    extreme event count, and average crop yield.
    Handles an unknown location gracefully instead of crashing.
    """
    available_locations = get_unique_locations(data)
    if location not in available_locations:
        return (f"Location not found.\n\nAvailable locations:\n"
                + "\n".join(available_locations))

    record_count = 0
    total_temperature = 0
    maximum_temperature = 0
    total_rainfall = 0
    maximum_rainfall = 0
    total_crop_yield = 0
    disruption_count = 0
    extreme_event_count = 0

    for record in data:
        if record["location"] == location:
            record_count += 1
            total_temperature += record["temperature"]
            total_rainfall += record["rainfall"]
            total_crop_yield += record["crop_yield"]

            if record["temperature"] > maximum_temperature:
                maximum_temperature = record["temperature"]
            if record["rainfall"] > maximum_rainfall:
                maximum_rainfall = record["rainfall"]
            if record["disruption"] == "Yes":
                disruption_count += 1
            if is_extreme_event(record):
                extreme_event_count += 1

    average_temperature = total_temperature / record_count
    average_crop_yield = total_crop_yield / record_count

    return f"""
=========================================
LOCATION WEATHER ANALYSIS
=========================================
Location: {location}
Records Analysed: {record_count}
Average Temperature: {average_temperature:.1f}°C
Maximum Temperature: {maximum_temperature}°C
Total Rainfall: {total_rainfall}mm
Maximum Rainfall: {maximum_rainfall}mm
Average Crop Yield: {average_crop_yield:.2f}
Operational Disruptions: {disruption_count}
Extreme Weather Events: {extreme_event_count}
"""


# ================================================================
# SECTION 8: RISK SCORING (Tasks 16-17)
# ================================================================

def calculate_risk_score(rainfall, temperature, disruption):
    """
    Calculate a weather risk score out of a rainfall score, a temperature
    score, and a disruption score, using the project's scoring model.
    Rainfall:    No/Light=0, Moderate=1, Heavy=2, Extreme=4
    Temperature: Cool/Moderate=0, Hot=2, Extreme Heat=4
    Disruption:  No=0, Yes=3
    """
    if rainfall <= 10:
        rainfall_score = 0
    elif rainfall <= 50:
        rainfall_score = 1
    elif rainfall <= 100:
        rainfall_score = 2
    else:
        rainfall_score = 4

    # Aligned with classify_temperature: Cool/Moderate = <=30, Hot = 31-35, Extreme = >35
    if temperature <= 30:
        temperature_score = 0
    elif temperature <= 35:
        temperature_score = 2
    else:
        temperature_score = 4

    if disruption == "No":
        disruption_score = 0
    else:
        disruption_score = 3

    return rainfall_score + temperature_score + disruption_score


def classify_risk(score):
    """
    Classify a risk score into a risk level:
    0-2 = Low | 3-5 = Moderate | 6-8 = High | 9+ = Critical
    """
    if score <= 2:
        return "Low Risk"
    elif score <= 5:
        return "Moderate Risk"
    elif score <= 8:
        return "High Risk"
    else:
        return "Critical Risk"


# ================================================================
# SECTION 9: WEATHER RISK ALERTS (Task 18)
# ================================================================

def generate_weather_alert(location, date, temperature, rainfall, wind_speed, disruption):
    """
    Generate a full risk alert for a single observation, combining the
    reason-based alert with the numeric risk score and risk level.
    """
    reasons = []
    if rainfall > 100:
        reasons.append("Extreme rainfall")
    if temperature > 35:
        reasons.append("Extreme temperature")
    if wind_speed > 25:
        reasons.append("Extreme wind speed")

    # Compare case-insensitively and turn it into a real boolean,
    # so "No" is never accidentally treated as truthy.
    has_disruption = disruption.strip().lower() == "yes"

    score = calculate_risk_score(rainfall, temperature, "Yes" if has_disruption else "No")
    risk_level = classify_risk(score)

    reason_block = ""
    if reasons:
        reason_string = ", ".join(reasons)
        disruption_string = " alongside operational disruption." if has_disruption else "."
        reason_block = f"Reason:\n{reason_string}{disruption_string}"

    return f"""========================================
⚠ WEATHER RISK ALERT
========================================
Location: {location}
Date: {date}
Temperature: {temperature}°C
Rainfall: {rainfall}mm
Wind Speed: {wind_speed}km/h
Disruption: {"Yes" if has_disruption else "No"}
Risk Score: {score}
Risk Level: {risk_level}

{reason_block}
"""


# ================================================================
# SECTION 10: LOCATION COMPARISON (Task 19)
# ================================================================

def compare_locations(location_1, location_2, data):
    """
    Compare two locations on average temperature, maximum temperature,
    total rainfall, extreme events, disruptions, average crop yield,
    and average risk score, then produce a simple conclusion.
    """
    available_locations = get_unique_locations(data)
    if location_1 not in available_locations or location_2 not in available_locations:
        return ("One of the locations entered was not found.\n\nAvailable locations:\n"
                + "\n".join(available_locations))

    stats = {}
    for location in (location_1, location_2):
        record_count = 0
        total_temperature = 0
        maximum_temperature = 0
        total_rainfall = 0
        total_crop_yield = 0
        disruption_count = 0
        extreme_event_count = 0
        total_risk = 0

        for record in data:
            if record["location"] == location:
                record_count += 1
                total_temperature += record["temperature"]
                total_rainfall += record["rainfall"]
                total_crop_yield += record["crop_yield"]

                if record["temperature"] > maximum_temperature:
                    maximum_temperature = record["temperature"]
                if record["disruption"] == "Yes":
                    disruption_count += 1
                if is_extreme_event(record):
                    extreme_event_count += 1

                total_risk += calculate_risk_score(
                    record["rainfall"], record["temperature"], record["disruption"]
                )

        stats[location] = {
            "average_temperature": total_temperature / record_count,
            "maximum_temperature": maximum_temperature,
            "total_rainfall": total_rainfall,
            "average_crop_yield": total_crop_yield / record_count,
            "disruption_count": disruption_count,
            "extreme_event_count": extreme_event_count,
            "average_risk": total_risk / record_count,
        }

    a = stats[location_1]
    b = stats[location_2]

    greater_heat = location_1 if a["average_temperature"] > b["average_temperature"] else location_2
    greater_rainfall = location_1 if a["total_rainfall"] > b["total_rainfall"] else location_2
    greater_risk = location_1 if a["average_risk"] > b["average_risk"] else location_2

    return f"""========================================
LOCATION COMPARISON
========================================
{location_1}
Average Temperature: {a['average_temperature']:.1f}°C
Maximum Temperature: {a['maximum_temperature']}°C
Total Rainfall: {a['total_rainfall']}mm
Extreme Events: {a['extreme_event_count']}
Disruptions: {a['disruption_count']}
Average Crop Yield: {a['average_crop_yield']:.2f}
Average Risk Score: {a['average_risk']:.1f}

{location_2}
Average Temperature: {b['average_temperature']:.1f}°C
Maximum Temperature: {b['maximum_temperature']}°C
Total Rainfall: {b['total_rainfall']}mm
Extreme Events: {b['extreme_event_count']}
Disruptions: {b['disruption_count']}
Average Crop Yield: {b['average_crop_yield']:.2f}
Average Risk Score: {b['average_risk']:.1f}

KEY OBSERVATION:
{greater_heat} shows greater heat exposure.
{greater_rainfall} shows greater rainfall exposure.
{greater_risk} shows the higher average weather risk.
"""


# ================================================================
# SECTION 11: NESTED LOOPS - LOCATION LEADERBOARD (Task 20)
# Also answers final business questions 3, 4, 5, 6 and 9.
# ================================================================

def location_leaderboard(data):
    """
    Demonstrates nested loops: for each location, loop through every
    weather record, check whether it belongs to that location, and if
    so, fold it into that location's running totals.

    Along the way this identifies which location leads on average
    temperature, total rainfall, extreme events, disruptions, and risk.
    """
    locations = get_unique_locations(data)

    leaders = {
        "average_temperature": ("", 0),
        "total_rainfall": ("", 0),
        "extreme_event_count": ("", 0),
        "disruption_count": ("", 0),
        "average_risk": ("", 0),
    }

    print("""
========================================
LOCATION LEADERBOARD
========================================""")

    # Outer loop: one pass per location
    for location in locations:
        record_count = 0
        total_temperature = 0
        total_rainfall = 0
        extreme_event_count = 0
        disruption_count = 0
        total_risk = 0

        # Inner loop: check every record and analyse the ones that match
        for record in data:
            if record["location"] == location:
                record_count += 1
                total_temperature += record["temperature"]
                total_rainfall += record["rainfall"]
                if is_extreme_event(record):
                    extreme_event_count += 1
                if record["disruption"] == "Yes":
                    disruption_count += 1
                total_risk += calculate_risk_score(
                    record["rainfall"], record["temperature"], record["disruption"]
                )

        average_temperature = total_temperature / record_count
        average_risk = total_risk / record_count

        print(f"{location}: Avg Temp {average_temperature:.1f}°C | "
              f"Total Rainfall {total_rainfall}mm | "
              f"Extreme Events {extreme_event_count} | "
              f"Disruptions {disruption_count} | "
              f"Avg Risk {average_risk:.1f} ({classify_risk(round(average_risk))})")

        if average_temperature > leaders["average_temperature"][1]:
            leaders["average_temperature"] = (location, average_temperature)
        if total_rainfall > leaders["total_rainfall"][1]:
            leaders["total_rainfall"] = (location, total_rainfall)
        if extreme_event_count > leaders["extreme_event_count"][1]:
            leaders["extreme_event_count"] = (location, extreme_event_count)
        if disruption_count > leaders["disruption_count"][1]:
            leaders["disruption_count"] = (location, disruption_count)
        if average_risk > leaders["average_risk"][1]:
            leaders["average_risk"] = (location, average_risk)

    print(f"""
KEY FINDINGS:
Highest average temperature: {leaders["average_temperature"][0]} ({leaders["average_temperature"][1]:.1f}°C)
Highest total rainfall: {leaders["total_rainfall"][0]} ({leaders["total_rainfall"][1]}mm)
Most extreme events: {leaders["extreme_event_count"][0]} ({leaders["extreme_event_count"][1]})
Most disruptions: {leaders["disruption_count"][0]} ({leaders["disruption_count"][1]})
Highest calculated risk: {leaders["average_risk"][0]} (avg score {leaders["average_risk"][1]:.1f})
""")

    return leaders


# ================================================================
# SECTION 12: MONTHLY ANALYSIS (Task 21)
# ================================================================

def analyse_month(data, target_month):
    """
    Analyse all records for one month. Returns None if the month is
    not represented in the dataset, so the caller can handle it.
    """
    total_temperature = 0
    total_rainfall = 0
    total_yield = 0
    extreme_heat_count = 0
    extreme_rainfall_count = 0
    disruption_days = 0
    record_count = 0

    for record in data:
        if get_month_name(record["date"]) == target_month:
            record_count += 1
            total_temperature += record["temperature"]
            total_rainfall += record["rainfall"]
            total_yield += record["crop_yield"]
            if record["temperature"] > 35:
                extreme_heat_count += 1
            if record["rainfall"] > 100:
                extreme_rainfall_count += 1
            if record["disruption"] == "Yes":
                disruption_days += 1

    if record_count == 0:
        return None

    return {
        "month": target_month,
        "record_count": record_count,
        "average_temperature": total_temperature / record_count,
        "total_rainfall": total_rainfall,
        "extreme_heat_count": extreme_heat_count,
        "extreme_rainfall_count": extreme_rainfall_count,
        "disruption_days": disruption_days,
        "average_yield": total_yield / record_count,
    }


def monthly_analysis_report(data):
    """
    Run analyse_month() for every month present in the dataset, print
    each month's summary, and identify the most weather-sensitive month
    (the month with the most combined extreme heat, extreme rainfall,
    and disruption days).
    """
    months = get_unique_months(data)
    most_sensitive_month = ""
    highest_sensitivity_score = -1

    print("""
========================================
MONTHLY WEATHER ANALYSIS
========================================""")

    for month in months:
        result = analyse_month(data, month)
        sensitivity_score = (result["extreme_heat_count"]
                              + result["extreme_rainfall_count"]
                              + result["disruption_days"])

        print(f"""
Month: {result['month']}
Records Analysed: {result['record_count']}
Average Temperature: {result['average_temperature']:.2f}°C
Total Rainfall: {result['total_rainfall']}mm
Extreme Heat Events: {result['extreme_heat_count']}
Extreme Rainfall Events: {result['extreme_rainfall_count']}
Total Disruption Days: {result['disruption_days']}
Average Crop Yield: {result['average_yield']:.2f}""")

        if sensitivity_score > highest_sensitivity_score:
            highest_sensitivity_score = sensitivity_score
            most_sensitive_month = month

    print(f"\nMost weather-sensitive month: {most_sensitive_month} "
          f"(sensitivity score: {highest_sensitivity_score})")


# ================================================================
# SECTION 13: RAINFALL vs DISRUPTION (Task 22)
# ================================================================

def analyse_rainfall_disruption(data):
    """
    Compare average rainfall and wind speed on days with disruption vs
    days without, to look for an association. Does not claim causation.
    """
    total_rainfall_disrupted = 0
    total_wind_disrupted = 0
    count_disrupted = 0
    total_rainfall_normal = 0
    total_wind_normal = 0
    count_normal = 0

    for record in data:
        if record["disruption"] == "Yes":
            count_disrupted += 1
            total_rainfall_disrupted += record["rainfall"]
            total_wind_disrupted += record["wind_speed"]
        else:
            count_normal += 1
            total_rainfall_normal += record["rainfall"]
            total_wind_normal += record["wind_speed"]

    average_rainfall_disrupted = total_rainfall_disrupted / count_disrupted
    average_rainfall_normal = total_rainfall_normal / count_normal
    average_wind_disrupted = total_wind_disrupted / count_disrupted
    average_wind_normal = total_wind_normal / count_normal

    print(f"""
========================================
RAINFALL vs OPERATIONAL DISRUPTION
========================================
Disrupted days ({count_disrupted} records):
  Average Rainfall: {average_rainfall_disrupted:.1f}mm
  Average Wind Speed: {average_wind_disrupted:.1f}km/h

Non-disrupted days ({count_normal} records):
  Average Rainfall: {average_rainfall_normal:.1f}mm
  Average Wind Speed: {average_wind_normal:.1f}km/h

The analysis identifies an association between periods of higher rainfall
and increased operational disruption in this sample. Further investigation
would be required to establish causality.
""")


# ================================================================
# SECTION 14: TEMPERATURE vs CROP YIELD (Task 23)
# ================================================================

def analyse_temperature_yield(data):
    """
    Compare average crop yield for hot/extreme-heat records vs
    cool/moderate records, to look for an association. Does not
    claim causation.
    """
    total_yield_hot = 0
    count_hot = 0
    total_yield_moderate = 0
    count_moderate = 0

    for record in data:
        category = classify_temperature(record["temperature"])
        if category in ("Hot", "Extreme Heat"):
            total_yield_hot += record["crop_yield"]
            count_hot += 1
        else:
            total_yield_moderate += record["crop_yield"]
            count_moderate += 1

    average_yield_hot = total_yield_hot / count_hot
    average_yield_moderate = total_yield_moderate / count_moderate

    print(f"""
========================================
TEMPERATURE vs CROP YIELD
========================================
Hot / Extreme Heat records ({count_hot} records):
  Average Crop Yield: {average_yield_hot:.2f}

Cool / Moderate records ({count_moderate} records):
  Average Crop Yield: {average_yield_moderate:.2f}

Within this simulated dataset, higher temperatures are observed alongside
lower crop-yield values, notably in Kano. Further investigation would be
required to determine whether temperature is a causal factor.
""")


# ================================================================
# SECTION 15: MANAGEMENT RECOMMENDATIONS (Task 26)
# ================================================================

def display_recommendations(data):
    """
    Present findings using the Observation / Interpretation /
    Recommendation structure, separating what the data shows from
    what it might mean and what management could consider doing.
    """
    leaders = location_leaderboard(data)
    highest_risk_location = leaders["average_risk"][0]

    print(f"""
========================================
MANAGEMENT RECOMMENDATIONS
========================================

1) RAINFALL AND OPERATIONAL DISRUPTION
Observation: Disrupted days show a higher average rainfall than
non-disrupted days in this sample.
Interpretation: This suggests heavy rainfall periods are associated with
a greater likelihood of operational disruption.
Recommendation: Consider contingency planning (e.g. flexible scheduling
or logistics buffers) around forecast periods of heavy rainfall.

2) TEMPERATURE AND CROP YIELD
Observation: Kano recorded the highest temperatures in the sample,
alongside its lowest crop-yield values.
Interpretation: The sample indicates greater heat exposure in Kano may
be linked to reduced agricultural output, though this is not proven
causally with this dataset.
Recommendation: Organisations operating in heat-sensitive environments
like Kano could explore heat-related agricultural contingency planning.

3) HIGHEST-RISK LOCATION
Observation: {highest_risk_location} recorded the highest average
weather risk score in this sample.
Interpretation: {highest_risk_location} may face the greatest combined
exposure to rainfall, temperature, and disruption-related risk factors.
Recommendation: Prioritise {highest_risk_location} for closer monitoring
and early-warning attention relative to the other locations studied.

LIMITATIONS: This is a simulated educational dataset with only 30
observations across 3 locations. Findings are associations only, not
proof of causation, and should not be used for real operational decisions.
""")


# ================================================================
# SECTION 16: INTERACTIVE MENU
# ================================================================

def run_menu():
    """Run the interactive Weather Impact Analysis System menu."""
    choice = 0
    while choice != 12:
        print("""
============================================
 WEATHER IMPACT ANALYSIS SYSTEM
============================================
1. Dataset Summary
2. View All Records
3. Highest Temperature
4. Highest Rainfall
5. Analyse Location
6. View Extreme Events
7. Generate Risk Alert
8. Compare Locations
9. Location Leaderboard (Nested Loops)
10. Monthly Analysis
11. Recommendations
12. Exit""")

        try:
            choice = int(input("\nEnter your choice: "))
        except ValueError:
            print("Invalid input! Please enter a number between 1 and 12.")
            continue

        print()

        if choice == 1:
            display_dataset_summary(weather_data)
            print("Average Temperature (all records):",
                  round(calculate_average_temperature(weather_data), 1), "°C")
            print("Total Rainfall (all records):", calculate_total_rainfall(weather_data), "mm")
            print("Total Disruptions (all records):", count_disruptions(weather_data))

        elif choice == 2:
            for record in weather_data:
                print(f"{record['date']} | {record['location']} | "
                      f"{record['temperature']}°C | {record['rainfall']}mm")

        elif choice == 3:
            temperature, date, location = find_highest_temperature(weather_data)
            print(f"Highest Temperature: {temperature}°C on {date} in {location}")

        elif choice == 4:
            rainfall, date, location = find_highest_rainfall(weather_data)
            print(f"Highest Rainfall: {rainfall}mm on {date} in {location}")

        elif choice == 5:
            location_input = input("Enter location (Lagos, Kano, Ibadan): ").strip().title()
            print(analyse_location(location_input, weather_data))

        elif choice == 6:
            display_extreme_events(weather_data)

        elif choice == 7:
            risk_location = input("Enter location: ").strip().title()
            risk_date = input("Enter date (YYYY-MM-DD): ").strip()
            risk_temperature = int(input("Enter temperature (°C): "))
            risk_rainfall = int(input("Enter rainfall (mm): "))
            risk_wind_speed = int(input("Enter wind speed (km/h): "))
            risk_disruption = input("Disruption (Yes/No): ").strip()
            print(generate_weather_alert(risk_location, risk_date, risk_temperature,
                                          risk_rainfall, risk_wind_speed, risk_disruption))

        elif choice == 8:
            location_1 = input("Enter first location: ").strip().title()
            location_2 = input("Enter second location: ").strip().title()
            print(compare_locations(location_1, location_2, weather_data))

        elif choice == 9:
            location_leaderboard(weather_data)

        elif choice == 10:
            monthly_analysis_report(weather_data)
            analyse_rainfall_disruption(weather_data)
            analyse_temperature_yield(weather_data)

        elif choice == 11:
            display_recommendations(weather_data)

        elif choice == 12:
            print("Closing Weather Impact Analysis System.")

        else:
            print("Out of range! Please enter a number between 1 and 12.")


# ================================================================
# RUN THE PROGRAM
# ================================================================

if __name__ == "__main__":
    run_menu()
```
